### HYPOTHESIS 1
### Province Risk Difference
### KPI: Claim Frequency

```python
province_results = run_hypothesis_test(
    df=df,
    hypothesis_name='Province Risk Difference',
    group_col='province',
    target_col='totalclaims',
    group_a='Oromia',
    group_b='Tigray',
    test_type='chi_square'
)

print(province_results)
```

### HYPOTHESIS 2
### Zip Code Risk Difference
### KPI: Claim Severity

```python
zipcode_risk_results = run_hypothesis_test(
    df=df,
    hypothesis_name='Zip Code Risk Difference',
    group_col='zipcode',
    target_col='ClaimSeverity',
    group_a=2000,
    group_b=8001,
    test_type='t_test'
)

print(zipcode_risk_results)
```

### HYPOTHESIS 3
### Margin Difference Between Zip Codes
### KPI: Margin

```python
margin_results = run_hypothesis_test(
    df=df,
    hypothesis_name='Margin Difference',
    group_col='zipcode',
    target_col='Margin',
    group_a=2000,
    group_b=8001,
    test_type='t_test'
)

print(margin_results)
```

### HYPOTHESIS 4
### Gender Risk Difference
### KPI: Claim Frequency

```python
gender_results = run_hypothesis_test(
    df=df,
    hypothesis_name='Gender Risk Difference',
    group_col='gender',
    target_col='HasClaim',
    group_a='Male',
    group_b='Female',
    test_type='chi_square'
)

print(gender_results)
```
